```python
#As a dict

class Person:

  def __init__(self,fname,lname,age,gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

person = Person('Avinash','Oraon',21,'Male')

import json
def show_object(person):
  if isinstance(person,Person):
    return {
        'fname':person.fname,'lname':person.lname,'age':person.age,'gender':person.gender}

  with open('demo.json','wb') as f:
    json.dump(person,f,default=show_object, indent = 4)

```

```python
with open('demo.json','rb') as f:
  data = json.load(f)
  print(data)
```

Pickling is a process whereby a Python object hierarchy is converted into a byte stream, and unpickling is the inverse operation,whereby a byte system(from a binary file or bytes-like object) is converted back into an object hierarchy

```python
class Person:

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('hi my name is', self.name,'and i am', self.age,'years old')
```

```python
p = Person('Avinash',21)
```

```python
import pickle
with open('person.pkl','wb') as f:
  pickle.dump(p,f)
```

```python
import pickle
with open('person.pkl','rb') as f:
  p = pickle.load(f)
p.display_info()
```

Pickle lets the user to store data in binary format. JSON lets the user store data in a human-readable text format
