Pandas series is like a column in a table . It is a 1-d array holding data of any type

```python
import numpy as np
import pandas as pd
```

```python
country = ['India','Pakistan','USA','Nepal','Srilanka']
pd.Series(country)
```

```python
runs = [13,24,56,78,100]
runs_series =pd.Series(runs)
```

```python
#custom index
marks = [67,57,89,100]
subjects = ['Maths','Science','English','Hindi']
pd.Series(marks,index=subjects,name = 'Avinash ke Marks')
```

```python
marks = {
    'Maths':67,
    'Science':57,
    'English':89,
    'Hindi':100
}
marks_series = pd.Series(marks, name = 'Avinash ke marks')
marks_series
```
