```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
X_train_tensor = torch.from_numpy(X_train.astype(np.float32))
X_test_tensor = torch.from_numpy(X_test.astype(np.float32))
y_train_tensor = torch.from_numpy(y_train.astype(np.float32))
y_test_tensor = torch.from_numpy(y_test.astype(np.float32))
```

```python
X_train_tensor.shape, X_test_tensor.shape, y_train_tensor.shape, y_test_tensor.shape
```

```python
class CustomDataset(Dataset):

  def __init__(self,features,labels):

    self.features = features
    self.labels = labels

  def __len__(self):
    return self.features.shape[0]

  def __getitem__(self,idx):
    return self.features[idx],self.labels[idx]
```

```python
train_dataset = CustomDataset(X_train_tensor, y_train_tensor)
test_dataset = CustomDataset(X_test_tensor, y_test_tensor)
```

```python
train_dataset = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_dataset = DataLoader(test_dataset, batch_size=32, shuffle=False)
```

```python
import torch.nn as nn

class MySimpleNN(nn.Module):

  def __init__(self,num_features):

    super().__init__()
    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out=  self.linear(features)
    out = self.sigmoid(out)
    return out
```

```python
learning_rate = 0.1
epochs = 25
```

```python
#create model
model = MySimpleNN(X_train_tensor.shape[1])

#define optimizer
optimizer = torch.optim.SGD(model.parameters(),lr=learning_rate)

#define loss function
loss_fn = nn.BCELoss()
```

```python
#define loop
for epoch in range(epochs):

  for batch_features ,batch_labels in train_dataset:

    y_pred = model(batch_features)

    loss = loss_fn(y_pred, batch_labels.view(-1,1))

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

  print(f'Epoch {epoch+1}/{epochs}, Loss: {loss.item():.4f}')

```

Evaluation

```python
model.eval()
accuracy_list = []

with torch.no_grad():
  for batch_features, batch_labels in test_dataset:
    y_pred = model(batch_features)
    y_pred_labels = (y_pred > 0.5).float()

    batch_accuracy = (y_pred.view(-1)== batch_labels).float().mean().item()
    accuracy_list.append(batch_accuracy)

  overall_accuracy = np.mean(accuracy_list)
  print(f'Accuracy: {overall_accuracy:.4f}')
```
