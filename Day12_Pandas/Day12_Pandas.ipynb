Pivot Table
The pivot table takes simple column-wise data as input, and groups the entries into a  two-dimensional table that provides a multidimensional summarization of the data

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
```

```python
df
```

```python
df.groupby(['sex','smoker'])[['total_bill']].mean().unstack()
```

```python
df.pivot_table(values='total_bill',index='sex',columns='smoker')
```

```python
df.pivot_table(values='total_bill',index='sex',columns='smoker',aggfunc='sum')
```

```python
#all cols together
df.pivot_table(values=['total_bill', 'tip', 'size'], index='sex',columns='smoker',aggfunc='sum')
```

```python
df.pivot_table(values=['total_bill', 'tip', 'size'], index=['sex','smoker'],columns=['day','time'],aggfunc='sum',margins= True)
```

```python
df
```

```python
#margins
df.pivot_table(values='total_bill',index='sex',columns='smoker',aggfunc='sum',margins = True)
```

```python
pd.date_range(start = '2023/1/5', end = '2023/2/10', freq = 'W') #Sunday
```

```python
pd.date_range(start = '2023/1/5', end = '2023/2/10', freq = 'W-THU') #AnyDay
```

```python
pd.date_range(start = '2023/1/5', end = '2023/1/7', freq = 'H') #Hourly
```

```python
pd.date_range(start = '2023/1/5', end = '2023/12/10', freq = 'M') #month end
```

```python
pd.date_range(start = '2023/1/5', end = '2023/12/10', freq = 'MS') #month start
```

```python
pd.date_range(start = '2023/1/5', end = '2026/12/10', freq = 'MS') #yearly
```

```python
#using periods
pd.date_range(start = '2023/1/5', periods = 10, freq = 'M')
```

to_datetime function
converts an existing object to pandas timestamp/datetimeindex object

```python
s = pd.Series(['2023/1/1','2022/1/1','2021/1/1'])
pd.to_datetime(s).dt.month_name()
#dt accessor
```

```python
#with errors
s = pd.Series(['2023/1/1','2022/1/1','2021/130/1'])
pd.to_datetime(s,errors = 'coerce') #corece
#NaT : Not a time
```
