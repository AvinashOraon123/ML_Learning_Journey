```python
b + 5
```

```python
import torch
import time

size = 10000

```

```python
A = torch.rand(size, size)
B = torch.rand(size, size)

start = time.time()
C = torch.matmul(A, B)
end = time.time()
print(f"CPU time: {end - start:.4f} seconds")
```

```python
if torch.cuda.is_available():
    A_gpu = A.to(device)
    B_gpu = B.to(device)

    # Warm-up run
    torch.matmul(A_gpu, B_gpu)

    start = time.time()
    C_gpu = torch.matmul(A_gpu, B_gpu)
    end = time.time()
    print(f"GPU time: {end - start:.4f} seconds")
else:
    print("CUDA is not available. Cannot perform GPU comparison.")
```
