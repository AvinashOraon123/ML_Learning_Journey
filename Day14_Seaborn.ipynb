1. Relational Plot


*   to see the statistical relation between 2 or more variables
*  Bivariate Analysis



```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
#Scatterplot -> axes level function
tips = sns.load_dataset('tips')
sns.scatterplot(x='total_bill',y='tip',data=tips, hue = 'sex', style ='time')
plt.show()
```

```python
#scatter using relplot -> size and hue -> figure level function
sns.relplot(x='total_bill',y='tip',data=tips,size='size',hue='size',style = 'sex') #size - depending on the size it increases the point size
plt.show()
```

```python
#line plot
gap = px.data.gapminder()
temp_df = gap[gap['country'] == 'India']
temp_df
```

```python
#axes level function
sns.lineplot(data = temp_df, x = 'year', y = 'lifeExp')
plt.show()
```

```python
#using rel plot
sns.relplot(data = temp_df, x = 'year', y = 'lifeExp', kind = 'line')
plt.show()
```

```python
#hue -> style
temp_df = gap[gap['country'].isin(['India','China','Pakistan'])]
```

```python
sns.relplot(kind= 'line', data = temp_df, x = 'year', y = 'lifeExp', hue = 'country')
plt.show()
```

```python
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
```

```python
sns.relplot(kind= 'line', data = temp_df, x = 'year', y = 'lifeExp', hue = 'country',style ='continent')
plt.show()
```

```python
#facet plot -> uses row and col -> work with relplot
#it will not work with scatterplot and lineplot

sns.relplot(data = tips, x ='total_bill', y ='tip', kind = 'line',col ='smoker', row = 'sex')
```

```python
#col wrap
sns.relplot(data = gap, x= 'lifeExp', y = 'gdpPercap', kind = 'scatter', col = 'year', col_wrap = 4)
```

A second way to plot Facet plots -> FacetGrid

```python
#figure level -> relplot -> displot -> catplot ->lmplot
sns.catplot(data=tips, x='sex',y='total_bill',col='day',kind='box',row='time')
```

```python
g = sns.FacetGrid(data=tips, col='day', row='time')
g.map(sns.boxplot, 'sex', 'total_bill')
```

Plotting Pairwise Relationship(PairGrid vs Pairplot)

```python
sns.pairplot(data = iris, hue = 'species')
```

```python
#pairgrid
g = sns.PairGrid(iris, hue = 'species', vars = ['sepal_length', 'sepal_width'])

g.map_offdiag(sns.scatterplot)
g.map_lower(sns.kdeplot)
g.map_diag(sns.boxplot)
#g.map_upper(sns.boxplot)
```
