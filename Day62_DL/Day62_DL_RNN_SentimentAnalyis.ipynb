```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train.shape
```

```python
y_test[0]
```

```python
len(X_train[2])
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train[0]
```

```python
model = Sequential()
model.add(SimpleRNN(32,input_shape=(150,1),return_sequences=False))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer = 'adam',metrics = ['accuracy'])
```

```python
model.fit(X_train,y_train, epochs = 5, validation_data = (X_test,y_test))
```

```python
model2 = Sequential()
model2.add(Embedding(10000, 2, input_length=50))
model2.add(SimpleRNN(32, return_sequences=False))
model2.add(Dense(1, activation='sigmoid'))
model2.build(input_shape=(None, 50)) # Explicitly build the model with the expected input shape
model2.summary()
```

Let's break down the expected trainable parameters for `model2`:

*   **Embedding Layer**: With an `input_dim` of 10000 and `output_dim` of 2, this layer has `10000 * 2 = 20000` trainable parameters.
*   **SimpleRNN Layer**: This layer has `32` units. The input to this layer comes from the Embedding layer (which has `2` features per timestep). The parameters are calculated as `(input_features * units) + (units * units) + units` (for input weights, recurrent weights, and biases respectively). So, `(2 * 32) + (32 * 32) + 32 = 64 + 1024 + 32 = 1120` trainable parameters.
*   **Dense Layer**: This is a single output neuron (`1` unit) connected to the `32` outputs from the `SimpleRNN` layer. The parameters are `(input_features * units) + units` (for weights and bias). So, `(32 * 1) + 1 = 33` trainable parameters.

Therefore, the total expected trainable parameters for `model2` are `20000 + 1120 + 33 = 21153`.
