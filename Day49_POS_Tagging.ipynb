```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(u'Tesla is looking at buying U.S. startup for $6 million')
```

```python
doc.text
```

```python
doc[-1]
```

```python
doc[2].pos_
```

```python
doc[3].pos_
```

```python
spacy.explain('PROPN')
```

```python
doc[4].tag_
```

```python
spacy.explain('VBG')
```

```python
for word in doc:
  print(word.text,'--->', word.pos_,',', word.tag_, spacy.explain(word.tag_))
```

```python
doc2 = nlp(u'I left the room')
doc3 = nlp(u'to the left of the room')
for word in doc2:
  print(word.text,'--->', word.pos_,',', word.tag_, spacy.explain(word.tag_))
```

```python
for word in doc3:
  print(word.text,'--->', word.pos_,',', word.tag_, spacy.explain(word.tag_))
```
