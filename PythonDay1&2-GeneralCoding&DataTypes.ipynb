*args and **kwargs are special Python keywords that are used to pass the variable length of arguements to a function

```python
def multiply(*args):
  product = 1
  for i in args:
    product = product*i
  print(args)
  return product

multiply(1,2,3,4)
```

```python
def add(*args):
  sum = 0
  for i in args:
    sum = sum+i
  print(args)
  return sum

add(1,2,4,5)
```

```python
def display(**kwargs):
  for (key,value) in kwargs.items():
    print(key, '->', value)

display(india='delhi',srilanka='colombo',nepal='kathmandu')
```

```python
#Higher order function : Input/Return higher order function


def transform(f,L):
  output = []
  for i in L:
    output.append(f(i))

  print(output)

L = [1,2,3,4,5]
transform(lambda x: x**3,L)
```
