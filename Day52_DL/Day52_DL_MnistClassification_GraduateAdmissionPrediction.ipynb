```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
(X_train,y_train), (X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
X_train[0].shape
```

```python
X_test.shape
```

```python
y_train
```

```python
import matplotlib.pyplot as plt
plt.imshow(X_train[0])
```

```python
X_test = X_test/255
X_train = X_train/255
```

```python
model = Sequential()
model.add(Flatten(input_shape = (28,28)))
model.add(Dense(128, activation= 'relu'))
model.add(Dense(32, activation = 'relu'))
model.add(Dense(10, activation = 'softmax'))
```

```python
model.summary()
```

```python
model.compile(loss = 'sparse_categorical_crossentropy', optimizer='Adam', metrics = ['accuracy'])
```

```python
history = model.fit(X_train,y_train,epochs=50, validation_split = 0.2)
```

```python
y_prob = model.predict(X_test)
```

```python
y_pred = y_prob.argmax(axis = 1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.legend(['loss','val_loss'])
plt.show()
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.legend(['accuracy','val_accuracy'])
plt.show()
```

# Lecture - 13 Graduate Admission Prediction using ANN


```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("mohansacharya/graduate-admissions")

print("Path to dataset files:", path)
```

```python
import os

# List the contents of the directory to find the CSV file
file_name = [f for f in os.listdir(path) if f.endswith('.csv')][0]
csv_file_path = os.path.join(path, file_name)

df = pd.read_csv(csv_file_path)
```

```python
df
```

```python
df.duplicated().sum()
```

```python
df.drop('Serial No.', axis = 1, inplace = True)
```

```python
X = df.iloc[:,0:-1]
y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(7, activation = 'relu', input_dim = 7))
model.add(Dense(7, activation = 'relu'))
model.add(Dense(1, activation = 'linear'))
```

```python
model.summary()
```

```python
model.compile(loss = 'mean_squared_error', optimizer = 'Adam')
```

```python
from sklearn.utils import validation
history = model.fit(X_train, y_train, epochs =100, validation_split = 0.2)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test, y_pred)
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.legend(['loss','val_loss'])
plt.show()
```
