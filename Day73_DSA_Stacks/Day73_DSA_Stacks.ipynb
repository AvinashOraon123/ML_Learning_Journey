```python
class Node:

  def __init__(self, data):
    self.data = data
    self.next = None
```

```python
class Stack:

  def __init__(self):
    self.top = None

  def isempty(self):
    return self.top == None

  def push(self,value):
    if self.top == None:
      self.top = Node(value)
    else:
      new_top = Node(value)
      new_top.next = self.top
      self.top = new_top

  def traverse(self):

    temp = self.top
    while temp!= None:
      print(temp.data)
      temp = temp.next

  def peek(self):

    if self.isempty():
      print("Stack is empty")
    else:
      return self.top.data

  def pop(self):
    if self.isempty():
      print("Stack is empty")
    else:
      data = self.top.data
      self.top = self.top.next
      return data




```

```python
s.push(1)
s.push(2)
s.push(3)
s.push(4)
s.push(5)
```

```python
# Q) Reverse a string
def reverse_string(text):
  s = Stack()
  for i in text:
    s.push(i)

  res = ""

  while(not s.isempty()):
    res += s.pop()

  return res
```

```python
reverse_string('Avinash Oraon')
```

```python
# Text Editor
def text_editor(text, pattern):
  u = Stack()
  r = Stack()

  for i in text:
    u.push(i)

  for i in pattern:

    if i == 'u':
      r.push(u.pop())
    elif i == 'r':
      u.push(r.pop())

  res = ""
  while(not u.isempty()):
    res = u.pop()+ res

  return res
```

```python
text_editor('Avinash', 'uuurr')
```
