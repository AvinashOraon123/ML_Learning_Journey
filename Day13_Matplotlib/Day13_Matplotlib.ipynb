Annotations

```python
batters =  pd.read_csv('batter.csv')
```

```python
sample_df = batters.head(100).sample(25, random_state = 5)
sample_df
```

```python

plt.figure(figsize = (18,7))
plt.scatter(sample_df['avg'],sample_df['strike_rate'], s = sample_df['runs'])
plt.xlabel('avg')
plt.ylabel('strike_rate')
plt.title('avg vs strike_rate')

plt.axhline(130, color = 'red', linestyle = '--')
plt.axvline(30, color = 'red', linestyle = '--')
for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
  #plt.annotate(sample_df['batter'].values[i],(sample_df['avg'].values[i],sample_df['strike_rate'].values[i]))
plt.show()
```

Horizontal and Vertical Lines

```python
fig,ax = plt.subplots(figsize = (18,7))

ax.scatter(batters['avg'],batters['strike_rate'])
ax.set_xlabel('avg')
ax.set_ylabel('strike_rate')
ax.set_title('avg vs strike_rate')

ax.axhline(130, color = 'red', linestyle = '--')
fig.show()
```

By using this syntax we can plot side by side

```python
fig,ax = plt.subplots(nrows = 2, ncols =1,sharex=True,figsize = (18,7))
ax[0].scatter(batters['avg'],batters['strike_rate'],color = 'red')
ax[1].scatter(batters['avg'],batters['runs'])
ax[0].set_xlabel('avg')
ax[0].set_ylabel('strike_rate')
ax[0].set_title('avg vs strike_rate')

ax[1].set_xlabel('avg')
ax[1].set_ylabel('runs')
ax[1].set_title('avg vs runs')
```

```python
fig,ax = plt.subplots(nrows = 2, ncols =2,figsize = (18,18))
ax[0,0].scatter(batters['avg'],batters['strike_rate'],color = 'red')
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'],color = 'red')
ax[1,1].hist(batters['runs'])
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(211)
ax1.scatter(batters['avg'],batters['strike_rate'],color = 'red')

ax2 = fig.add_subplot(212) #2,1,2 -> rows,column,graph no.
ax2.scatter(batters['avg'],batters['runs'])
```

Surface Plots

```python
x  = np.linspace(-10,10,100)
y  = np.linspace(-10,10,100)

xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2

fig = plt.figure(figsize = (12,8))
ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize = (12,8))
ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx,yy,z,cmap = 'plasma')
fig.colorbar(p)
```

Contour PLots


```python
z = xx**2 + yy**2

fig = plt.figure(figsize = (12,8))
ax = plt.subplot()
p = ax.contour(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
z = xx**2 + yy**2

fig = plt.figure(figsize = (12,8))
ax = plt.subplot()
p = ax.contourf(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize = (12,8))
ax = plt.subplot()
p = ax.contourf(xx,yy,z,cmap = 'plasma')
fig.colorbar(p)
```
