```python
import tensorflow as tf
from tensorflow import keras
from keras.layers import Dense, Conv2D, Flatten, MaxPooling2D, AveragePooling2D
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
model = Sequential()

model.add(Conv2D(32, (3,3),padding = 'valid', input_shape=(28, 28, 1), activation='relu'))
model.add(MaxPooling2D(pool_size = (2,2), strides = 2))

model.add(Conv2D(32, (3,3), padding = 'valid', activation='relu'))
model.add(MaxPooling2D(pool_size = (2,2), strides = 2))

model.add(Flatten())

model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.summary()
```

LeNET architecture

```python
model1 = Sequential()

model1.add(Conv2D(6, (5,5),padding = 'valid', input_shape=(32, 32, 1), activation='tanh'))
model1.add(AveragePooling2D(pool_size = (2,2), strides = 2))

model1.add(Conv2D(16, (5,5), padding = 'valid', activation='tanh'))
model1.add(AveragePooling2D(pool_size = (2,2), strides = 2))

model1.add(Flatten())

model1.add(Dense(120, activation='tanh'))
model1.add(Dense(84, activation='tanh'))
model1.add(Dense(10, activation='softmax'))
```
